```python
import pandas as pd
import seaborn as sns
```

```python
car = sns.load_dataset('car_crashes')
```

```python
car.head(15)
```

```python
titanic = pd.read_csv('titanic.csv')
```

```python
netflix= pd.read_csv('netflix_titles.csv')
```

1. Scatterplot(Numerical-Numerical)

```python
sns.scatterplot(x=car['speeding'], y=car['alcohol'], hue=car['total'], style=car['ins_losses'], size=car['abbrev'])
```

2. Boxplot(Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(x=titanic['Pclass'], y=titanic['Age'], hue=titanic['Sex'])
```

3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(x=titanic['Sex'], y=titanic['Age'], hue=titanic['Survived'])
```

4. Distplot (Numerical - Categorical)


```python
netflix.head(10)
```

```python
sns.kdeplot(netflix[netflix['type'] == 'Movie']['release_year'], label='Movie')
sns.kdeplot(netflix[netflix['type'] == 'TV Show']['release_year'], label='TV Show')
```

5. HeatMap (Categorical - Categorical)

```python
netflix.head(3)
```

```python
sns.heatmap(pd.crosstab(netflix['type'],netflix['rating']))
```

```python
titanic.groupby('Embarked')['Survived'].mean()*100
```

6. ClusterMap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(netflix['type'],netflix['rating']))
```
